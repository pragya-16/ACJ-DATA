```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin
```

```python
url = "https://quotes.toscrape.com/"

print(url)
```

```python
response = requests.get(url)

print(response.status_code)
```

```python
html = response.text

print(html[:1000])
```

```python
soup = BeautifulSoup(html, "html.parser")

print(type(soup))
```

```python
first_quote = soup.select_one("div.quote")

print(first_quote)
```

```python
quote_text = first_quote.select_one("span.text").get_text(strip=True)

print(quote_text)
```

```python
author = first_quote.select_one("small.author").get_text(strip=True)

print(author)
```

```python
tags = first_quote.select("div.tags a.tag")

print(tags)
```

```python
author_link = first_quote.select_one("small.author + a")

print(author_link)
```

```python
relative_author_url = author_link.get("href")

print(relative_author_url)
```

```python
tags = first_quote.select("div.tags a.tag")

print(tags)
```

```python
tag_names = [tag.get_text(strip=True) for tag in tags]

print(tag_names)
```

```python
author_link = first_quote.select_one("small.author + a")

print(author_link)
```

```python
author_url = urljoin(url, relative_author_url)

print(author_url)
```

```python
first_quote_data = {
    "quote": quote_text,
    "author": author,
    "tags": tag_names,
    "author_url": author_url
}

print(first_quote_data)
```

```python
quotes = soup.select("div.quote")

print(len(quotes))
```

```python
rows = []

for quote in quotes:

    quote_text = quote.select_one("span.text").get_text(strip=True)

    author = quote.select_one("small.author").get_text(strip=True)

    tags = quote.select("div.tags a.tag")
    tag_names = [tag.get_text(strip=True) for tag in tags]

    author_link = quote.select_one("small.author + a")
    relative_author_url = author_link.get("href")
    author_url = urljoin(url, relative_author_url)

    row = {
        "quote": quote_text,
        "author": author,
        "tags": tag_names,
        "author_url": author_url
    }

    rows.append(row)
```

```python
print(len(rows))
```

```python
df = pd.DataFrame(rows)

df.head()
```
